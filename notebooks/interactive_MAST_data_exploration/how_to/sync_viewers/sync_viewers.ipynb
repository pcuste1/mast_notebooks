## What is a Viewport?
In this tutorial, Viewport is defined by the [Astro Image Dislay API (AIDA)](https://github.com/astropy/astro-image-display-api/) specification as the set of parameters describing the state of the image being displayed by the widget. 

Both the `jdaviz` and `mast-aladin` widgets implement AIDA and define their viewports as the following set of properties:
- **Center**: The center of the viewport in Sky Coordinates
- **FOV**: The field of view (FOV) of the viewport. For viewers that are not square, the FOV is interpreted as the size of the shorter side of the viewer such that the FOV is guaranteed to be entirely visible regardless of the aspect ratio of the viewer.
- **Rotation**: The rotation applied to the image in the viewport
- **Projection**: The current projection applied to the viewport

### Imports 
For this tutorial, we require imports from the `jdaviz`, `mast-aladin` and `sidecar` packages 
- `ViewerSyncUI`: creates the UI which powers viewer syncing

```python
import jdaviz as jd

from mast_aladin import MastAladin
from mast_aladin.adapters import ViewerSyncUI

from sidecar import Sidecar
```

## Widget Setup
Before we can do any syncing between widgets, first we have to setup or jupyter notebook with widgets to sync.

**Initialize widgets:**
First, create an instance of `mast-aladin` and `jdaviz`. Each requires some setup to ensure they are looking at the same sources, so that syncing makes sense.

- mast-aladin: Set the target of `mast-aladin` to the Cartwheel Galaxy 
- imviz: Download a JWST nircam image of the Cartwheel Galaxy and load into `jdaviz`

```python
ma_viewer = MastAladin(target="Cartwheel Galaxy", fov=1, height=500)

uri = 'mast:JWST/product/jw02727-o002_t062_nircam_clear-f090w_i2d.fits'
jd.load(uri, format="Image", cache=True)
```

**Display:** The next setup step is to display the `mast_aladin` and `jdaviz` widgets using `sidecar`. For more information on how this is done, please see the `interactive_MAST_data_exploration/how_to/organize_widgets`. This step is not required, but makes working with this tool more plesant.

```python
from sidecar import Sidecar
# Create a sidecar
ma_sc = Sidecar(title="mast-aladin", anchor="split-right")

with ma_sc:
    display(ma_viewer)

# Create a sidecar with defined anchor
jd_sc = Sidecar(title="jdaviz Main", ref=ma_sc, anchor="split-bottom")
with jd_sc:
    jd.show()
```

## Viewer Sync UI

The Viewer Sync UI enables real-time synchronization between viewers. The UI provides the following configuration setttings:

- Source Widget: Select which viewer acts as the source of truth for the viewport (Imviz or Mast Aladin). Set to "None" to stop syncing.
- Properties: Choose which viewport properties to sync across viewers (Center, Field of View, Rotation, Projection).


```python
vsui = ViewerSyncUI()
vsui.display()
```
